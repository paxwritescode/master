```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.linestyle'] = '--'
plt.rcParams['grid.alpha'] = 0.6
```

```python
x = np.linspace(-10, 10, 500)
t_snapshots = [0, 1, 2] # Моменты времени для левого графика
time = np.linspace(0, 4, 200) # Временная сетка для правого графика

def intensity(x, t):
    return 2 * (np.tanh(x - 2 * t))**2

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

colors = ['blue', 'orange', 'green']
for t_val, col in zip(t_snapshots, colors):
    ax1.plot(x, intensity(x, t_val), label=f't = {t_val}', color=col, linewidth=2.5)

ax1.set_title('Dark Soliton Intensity Profiles over Time', fontsize=12, pad=10)
ax1.set_xlabel('Spatial Coordinate X', fontsize=11)
ax1.set_ylabel('Intensity $|u(x,y,t)|^2$', fontsize=11)
ax1.set_ylim(-0.1, 2.3)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(loc='lower right', fontsize=10)

X, T = np.meshgrid(x, time)
Z = intensity(X, T)

im = ax2.imshow(Z, extent=[-10, 10, 0, 4], origin='lower', aspect='auto', cmap='viridis')
cbar = fig.colorbar(im, ax=ax2)
cbar.set_label('Intensity $|u|^2$', fontsize=11)

ax2.set_title('Spatiotemporal Evolution of Dark Soliton', fontsize=12, pad=10)
ax2.set_xlabel('Spatial Coordinate X', fontsize=11)
ax2.set_ylabel('Time t', fontsize=11)

plt.suptitle('Dynamics of the Dark Soliton Solution for the NLS Equation', fontsize=15, fontweight='bold', y=0.98)

plt.tight_layout()
plt.subplots_adjust(top=0.85)

plt.savefig('results/dark_soliton_dynamics.pdf', dpi=300)
# plt.show()
plt.close()
```

```python
def real_part(x, t):
    return np.sqrt(2) * np.tanh(x - 2 * t) * np.cos(x - 2 * t)

def imag_part(x, t):
    return np.sqrt(2) * np.tanh(x - 2 * t) * np.sin(x - 2 * t)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

colors = ['blue', 'orange', 'green']
for t_val, col in zip(t_snapshots, colors):
    ax1.plot(x, real_part(x, t_val), label=f't = {t_val}', color=col, linewidth=1.8)
    ax2.plot(x, imag_part(x, t_val), label=f't = {t_val}', color=col, linewidth=1.8)

ax1.set_title('Real Part $\Re(u)$ of the Dark Soliton', fontsize=12, pad=10)
ax1.set_xlabel('Spatial Coordinate X', fontsize=11)
ax1.set_ylabel('$\Re(u(x,y,t))$', fontsize=11)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(loc='upper right', fontsize=10)

ax2.set_title('Imaginary Part $\Im(u)$ of the Dark Soliton', fontsize=12, pad=10)
ax2.set_xlabel('Spatial Coordinate X', fontsize=11)
ax2.set_ylabel('$\Im(u(x,y,t))$', fontsize=11)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend(loc='upper right', fontsize=10)

plt.suptitle('Oscillatory Behavior of the Dark Soliton (Real & Imaginary Components)', fontsize=15, fontweight='bold', y=0.98)

plt.tight_layout()
plt.subplots_adjust(top=0.85)

plt.savefig('results/dark_soliton_real_imag.pdf', dpi=300)
# plt.show()
plt.close()
```

```python
x_zoomed = np.linspace(-1, 0.5, 500)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
t_point = 1

ax1.plot(x_zoomed, real_part(x_zoomed, t_point), color='blue', linewidth=1.8)
ax2.plot(x_zoomed, imag_part(x_zoomed, t_point), color='orange', linewidth=1.8)

ax1.set_title('Real Part $\Re(u)$ of the Dark Soliton', fontsize=12, pad=10)
ax1.set_xlabel('Spatial Coordinate X', fontsize=11)
ax1.set_ylabel('$\Re(u(x,y,t))$', fontsize=11)
ax1.grid(True, linestyle=':', alpha=0.6)

ax2.set_title('Imaginary Part $\Im(u)$ of the Dark Soliton', fontsize=12, pad=10)
ax2.set_xlabel('Spatial Coordinate X', fontsize=11)
ax2.set_ylabel('$\Im(u(x,y,t))$', fontsize=11)
ax2.grid(True, linestyle=':', alpha=0.6)

plt.suptitle(f'Oscillatory Behavior of the Dark Soliton (Real & Imaginary Components), t = {t_point}', fontsize=15, fontweight='bold', y=0.98)

plt.tight_layout()
plt.subplots_adjust(top=0.85)

plt.savefig('results/dark_soliton_real_imag_zoomed.pdf', dpi=300)
plt.show()
plt.close()
```
