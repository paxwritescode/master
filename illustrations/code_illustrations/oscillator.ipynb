```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
omega = 2.0
y0 = 1.0
v0 = 0.0

t = np.linspace(0, 10, 1000)

y = y0 * np.cos(omega * t) + (v0 / omega) * np.sin(omega * t)
v = -y0 * omega * np.sin(omega * t) + v0 * np.cos(omega * t)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(t, y, label='Position $y(t)$', color='blue', linewidth=2)
ax1.plot(t, v, label='Velocity $y\'(t)$', color='orange', linestyle='--', linewidth=2)
ax1.set_title('Linear Harmonic Oscillator Dynamics', fontsize=12, pad=10) # Individual title 1
ax1.set_xlabel('Time $t$', fontsize=10)
ax1.set_ylabel('Values', fontsize=10)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(loc='upper right', fontsize=9)

ax2.plot(y, v, color='purple', linewidth=2)
ax2.set_title('Phase Portrait ($y$ vs $y\'$)', fontsize=12, pad=10) # Individual title 2
ax2.set_xlabel('Position $y$', fontsize=10)
ax2.set_ylabel('Velocity $y\'$', fontsize=10)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.axis('equal')
plt.suptitle('Harmonic Oscillator Model', fontsize=16, fontweight='bold', y=0.98)

plt.tight_layout()
plt.subplots_adjust(top=0.85)

plt.savefig('results/harmonic_oscillator.pdf', dpi=300)
# plt.show()
plt.close()
```
